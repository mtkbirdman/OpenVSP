# VSPAERO mesh optimization

`ThinWing / ThickWing / Hybrid / ThickAll` のいずれかを明示し、geometry・representation・clustering・wake 条件を固定したまま、`Tess_W` → `SectTess_U / Tess_U` の順に mesh convergence を確認する例です。


```python
from pathlib import Path
import sys

start_dir = Path.cwd().resolve()
repo_root = next((p for p in [start_dir, *start_dir.parents] if (p / 'src' / 'VSPAEROMesh.py').exists()), None)
if repo_root is None:
    raise FileNotFoundError('Could not find repository root.')
if str(repo_root) not in sys.path:
    sys.path.insert(0, str(repo_root))

from src.VSPAEROMesh import optimize_vspaero_tessellation
```

```python

model_name = "G103A"
model_path = repo_root / 'examples' / 'models' / f'{model_name}' / f'{model_name}.vsp3'
output_dir = repo_root / 'results' / f'mesh_optimization_{model_name}_Hybrid'

# ThinWing / ThickWing / Hybrid / ThickAll
representation = 'ThickAll'

result = optimize_vspaero_tessellation(
    input_vsp3_path=model_path,
    output_dir=output_dir,
    representation=representation,
    qoi_tolerances={
        'CLiw': 0.005,
        'CDiw': 0.01,
        'CMytot': 0.01,
    },
    qoi_scales={'CMytot': 0.05},
    tess_w_scales=(0.6, 1.0, 1.5),
    u_scales=(0.6, 1.0, 1.5),
    alpha=2.0,
    mach=0.1,
    reynolds_number=4.4e6,
    fixed_wake_flag=True,
    verbose=1,
)
```

```python
display(result['cases'])
print('fully_converged:', result['summary']['fully_converged'])
print('selected:', result['selected_vsp3_path'])
print('summary:', result['summary_json_path'])
```
