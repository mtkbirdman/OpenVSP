```python
from pathlib import Path
import sys

# pip install -e . is not needed: find src from the checkout.
start_dir = Path.cwd().resolve()
repo_root = next((p for p in [start_dir, *start_dir.parents] if (p / "src" / "openvsppy").is_dir()), None)
if repo_root is None:
    raise FileNotFoundError("Could not find repository root")
source_dir = str(repo_root / "src")
if source_dir not in sys.path:
    sys.path.insert(0, source_dir)
```

# VSPAERO Cp / Delta Cp 4-view capture
OpenVSP 3.50.5 の定常 `.adb` と対応する `.adb.cases` を使用します。


```python
from IPython.display import Image, display
from openvsppy.visualization.VSPAEROCpViewCapture import list_adb_cases, capture_adb_cp_views

adb_path = repo_root / "Aircraft.adb"  # 実際のファイルパスに変更
output_path = repo_root / "Aircraft_Cp.png"
for info in list_adb_cases(adb_path):
    print(info)
```

```python
image_path = capture_adb_cp_views(
    adb_path, output_path, case=1, size=(1600, 1200),
    cp_limits=(-1.5, 1.0), align_orthographic_views=True,
    overwrite=True,
)
display(Image(filename=str(image_path)))
```
