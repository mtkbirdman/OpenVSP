# OpenVSP model-list animation

順序付きの `.vsp3` リストを同じ画面設定で撮影し、MP4またはGIFへ結合します。モデルの読み込み順と重複はそのままフレーム順になります。OpenVSP GUIは撮影全体で1回だけ起動し、完了時に閉じます。

GUI対応のOpenVSP Python APIとPillowが必要です。MP4を作成する場合だけimageio-ffmpegも必要です。

```python
from pathlib import Path
import sys

from IPython.display import Video, display

start_dir = Path.cwd().resolve()
repo_root = next(
    (
        candidate
        for candidate in [start_dir, *start_dir.parents]
        if (candidate / "src" / "OpenVSPViewCapture.py").exists()
    ),
    None,
)
if repo_root is None:
    raise FileNotFoundError("Could not find the repository root.")
if str(repo_root) not in sys.path:
    sys.path.insert(0, str(repo_root))

from src.OpenVSPViewCapture import create_vsp3_animation  # noqa: E402

print("repo_root:", repo_root)
```

## Ordered input

ここでは同梱モデルを例にします。実際の計算ケースでは、このセルだけを `sorted(...)` や既存のcase一覧へ置き換えます。意図しない並べ替えを避けるため、関数内ではソートしません。

```python
vsp3_paths = [
    repo_root / "examples" / "models" / "SampleGlider" / "SampleGlider.vsp3",
    repo_root / "examples" / "models" / "G103A" / "G103A.vsp3",
    repo_root / "examples" / "models" / "Boeing_777-9x_mod" / "Boeing_777-9x_mod.vsp3",
]
vsp3_paths
```

## Capture and encode

出力拡張子を `.gif` に変えるとGIFになります。既定では連番PNGと撮影条件を `example_models_frames/` に保存します。同じ入力・画面設定で再実行した場合は、有効なフレームを再利用します。`fps` だけの変更では撮り直しません。

```python
output_dir = repo_root / "examples" / "notebooks" / "view_capture" / "output"
```

```python
result = create_vsp3_animation(
    vsp3_paths=vsp3_paths,
    output_path=output_dir / "example_models.mp4",
    size=(1200, 900),
    render_mode="hidden",
    fps=1,
    keep_frames=True,
    overwrite=True,
)
display(Video(str(result["animation_path"]), embed=True))
```

```python
from IPython.display import Image as IPythonImage, display

result = create_vsp3_animation(
    vsp3_paths=vsp3_paths,
    output_path=output_dir / "example_models.gif",
    size=(1200, 900),
    render_mode="hidden",
    fps=1,
    keep_frames=True,
    overwrite=True,
)
display(IPythonImage(filename=str(result["animation_path"])))
```
