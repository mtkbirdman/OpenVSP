# VSPAERO mesh-rule calibration dataset

This notebook appends raw calibration cases. It does not assign a good/bad mesh label.


```python
from pathlib import Path
import sys

start_dir = Path.cwd().resolve()
repo_root = next(
    (
        candidate
        for candidate in [start_dir, *start_dir.parents]
        if (candidate / "src" / "AnalysisVSPAERO.py").exists()
    ),
    None,
)

if repo_root is None:
    raise FileNotFoundError(
        "Could not find the repository root containing src/AnalysisVSPAERO.py. "
        "Run this notebook inside the repository or set repo_root manually."
    )

if str(repo_root) not in sys.path:
    sys.path.insert(0, str(repo_root))

from src.VSPAEROMeshDataset import build_vspaero_mesh_rule_dataset, rebuild_mesh_dataset_tables
```

```python
model_name = "Boeing_777-9x_mod"
model_path = repo_root / f'examples/models/{model_name}/{model_name}.vsp3'
dataset_dir = repo_root / 'examples/notebooks/mesh_rule_dataset/results'

result = build_vspaero_mesh_rule_dataset(
    input_vsp3_path=model_path,
    output_dir=dataset_dir,
    representation='ThickAll',
    lhs_samples=128,
    ncpu=8,
    wake_num_iter=None,
)
result['planned_case_count'], result['completed_this_run'], result['failed_this_run']
```
